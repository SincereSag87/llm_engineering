# Browser-based scraping demo

This notebook demonstrates the Playwright scraper on a JavaScript-heavy site. It does not use `.env` or any API keys.

Install the browser dependency once before running the notebook:

```bash
pip install -r requirements.txt
playwright install chromium
```

````python
from browser_scraper import BrowserScraperError, fetch_website_contents, fetch_website_links
````

````python
url = "https://openai.com"

try:
    contents = fetch_website_contents(url)
    print(contents)
except BrowserScraperError as exc:
    print(f"Scraping failed: {exc}")
````

````python
try:
    links = fetch_website_links(url)
    links[:10]
except BrowserScraperError as exc:
    print(f"Link scraping failed: {exc}")
````
